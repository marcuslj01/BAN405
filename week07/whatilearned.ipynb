Vectors can be used by:

```python
import numpy as np
import pandas as pd

prices_array = np.array([29.90, 35.00, 12.50, 22.70]) # this is a vector
quantities = np.array([3, 5, 0, 2]) # this is also a vector

revenue = prices_array * quantities # this is a new vector with the product of each item in the vector

print(revenue)
print(revenue.sum())
```

`float64` is a decimal number, `int64` is a whole number. If you build an array out of mixed
values, NumPy does not keep the mixture - it converts everything to a type that can hold all of it.

```python
print(np.array([1, 2, 3.5])) # 1 and 2 is turned into floats
print(np.array([1, 2, "3"])) #1 and 2 is turned into a string because of "3"
```

## Series & Dataframes
Pandas introduces "series" and "dataframes":
- A pandas **Series** is a NumPy array with a second thing attached: an **index**, giving every value a
label.


```python
emissions = pd.Series(
    [41.6, 434.3, 6.9, 12.4],
    index=["Norway", "Brazil", "Iceland", "Ghana"]
)

emissions

print(emissions.values)
print(emissions.values.dtype)
print(emissions.index)
```

- A **DataFrame** is a table: several Series side by side, all sharing one index. Building one from a
dictionary shows the structure clearly - each key becomes a column name.

```python
sales = pd.DataFrame({
    "product":  ["coffee", "tea", "juice", "sandwich"],
    "price":    [29.90, 35.00, 12.50, 48.00],
    "quantity": [3, 5, 0, 2],
})

sales

```

```python
print(sales.shape) # gives rows and columns amounts (4 rows, 3 cols in this case)
print(sales.columns)
print(sales.index)
```

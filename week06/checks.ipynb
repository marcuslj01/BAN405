I made a virtual environment called practice:
```
conda create --name practice python=3.14 pandas
```

Creating an environment does not start using it. Switch to it, and ask which version of pandas it has:

```
conda activate practice
```
```
conda list pandas
```

Btw: a venv is basically just a folder on the computer

Then activated it in the kernel. Now we can see what python version and pandas version we are using:

`````python
import sys
print(sys.executable)

import pandas
print(pandas.__version__)
`````

An environment can be exported as a .yml file, so another person can open the exact environment.

This exports the environment as a .yml file (from-history makes it only export the packages we explicitly ask for ):
```
conda env export --from-history > practice.yml 
````

Then we can deactivate and delete it:
```
conda deactivate
conda env remove --name practice
```

Now we can recreate the environment some other time, by using the following command in the correct directory:
```
conda env create --file practice.yml 
conda activate practice
```
